````python
# skip
# Temperature converter — Celsius to Fahrenheit and Kelvin
# Useful for normalising CinemaStream server temperature metrics
# that arrive in mixed units from regional data centres

# Collect input — wrap in float() because temperatures can be decimal
celsius = float(input("Enter temperature in Celsius: "))

# Standard conversion formulae
fahrenheit = celsius * 9 / 5 + 32
kelvin = celsius + 273.15  # Absolute zero offset

print(f"\n{celsius}°C is:")
print(f"  {fahrenheit:.2f}°F")
print(f"  {kelvin:.2f} K")
````

````python
# Regular class vs __slots__ class — the __dict__ difference

class RegularEvent:
    def __init__(self, movie_id, minutes):
        self.movie_id = movie_id
        self.minutes  = minutes

class SlottedEvent:
    __slots__ = ["movie_id", "minutes"]   # fixed attribute layout — no __dict__
    def __init__(self, movie_id, minutes):
        self.movie_id = movie_id
        self.minutes  = minutes

re = RegularEvent(101, 132)
se = SlottedEvent(101, 132)

# Shallow instance sizes look similar...
print(f"RegularEvent instance: {sys.getsizeof(re)} bytes")
print(f"SlottedEvent instance: {sys.getsizeof(se)} bytes")

# ...but RegularEvent carries a __dict__ that SlottedEvent does not
print(f"RegularEvent __dict__: {sys.getsizeof(re.__dict__)} bytes")

try:
    _ = se.__dict__
except AttributeError:
    print("SlottedEvent has no __dict__ — that overhead is gone")
````

````python
# ---- user_utils.py (preview) ----

def get_plan_price(plan: str) -> float:
    """Return the monthly price in SGD for a given plan."""
    prices = {"Free": 0.0, "Basic": 8.90, "Premium": 12.90}
    return prices.get(plan, 0.0)


def is_active(user: dict) -> bool:
    """Return True if user has not churned."""
    return not user.get("churned", True)


def user_label(user: dict) -> str:
    """
    Return a display label combining name, plan, and active status.
    
    Example: 'Ravi Kumar [Premium | active]'
    """
    status = "active" if is_active(user) else "churned"
    return f"{user['name']} [{user['plan']} | {status}]"


def compute_churn_rate(users: list) -> float:
    """Return churn rate as a percentage across a list of user dicts."""
    if not users:
        return 0.0
    churned = sum(1 for u in users if u.get("churned", False))
    return (churned / len(users)) * 100
````

### 2.1 Defining a Class

````python
class Counter:
    """A running total that you can increment, decrement, and reset."""

    def __init__(self, start: int = 0):
        # self.count is an INSTANCE attribute — each Counter has its own copy
        self.count = start

    def increment(self, by: int = 1) -> None:
        self.count += by

    def decrement(self, by: int = 1) -> None:
        self.count -= by

    def reset(self) -> None:
        self.count = 0

    def value(self) -> int:
        return self.count


c = Counter()       # start at 0
c.increment()       # count = 1
c.increment(5)      # count = 6
c.decrement(2)      # count = 4
print(c.value())    # 4

c2 = Counter(start=100)   # completely independent object
print(c2.value())         # 100 — c and c2 do NOT share state
````

### 2.2 Dunder Methods

````python
class Counter:
    def __init__(self, start: int = 0):
        self.count = start
        self._history = [start]   # underscore prefix = private by convention

    def increment(self, by: int = 1) -> None:
        self.count += by
        self._history.append(self.count)

    def __repr__(self) -> str:
        # __repr__: unambiguous; what the Python shell and repr() show
        return f"Counter(count={self.count})"

    def __str__(self) -> str:
        # __str__: human-friendly; what print() and str() show
        # falls back to __repr__ if not defined
        return f"counter is at {self.count}"

    def __len__(self) -> int:
        # len(c) → number of distinct states the counter has been through
        return len(self._history)

    def __eq__(self, other: object) -> bool:
        if not isinstance(other, Counter):
            return NotImplemented
        return self.count == other.count


c = Counter(10)
c.increment(5)   # count = 15
c.increment(3)   # count = 18

print(repr(c))            # Counter(count=18)
print(str(c))             # counter is at 18
print(len(c))             # 3  →  start=10, +5, +3 = three history entries
print(c == Counter(18))   # True
````

### 2.3 Inheritance

````python
class ReadOnlyCounter(Counter):
    """A snapshot of a counter: readable but frozen after creation."""

    def increment(self, by: int = 1) -> None:
        raise AttributeError("ReadOnlyCounter cannot be modified after creation")

    def decrement(self, by: int = 1) -> None:
        raise AttributeError("ReadOnlyCounter cannot be modified after creation")

    def reset(self) -> None:
        raise AttributeError("ReadOnlyCounter cannot be modified after creation")


snapshot = ReadOnlyCounter(start=42)
print(repr(snapshot))   # inherits __repr__ from Counter — no re-definition needed

try:
    snapshot.increment()
except AttributeError as exc:
    print(f"Blocked: {exc}")
````

### 2.4 `@property` — Computed Attributes

````python
class Temperature:
    """Stores temperature in Celsius; exposes both scales via @property."""

    def __init__(self, celsius: float):
        self._celsius = celsius   # underscore = private: callers use .celsius

    @property
    def celsius(self) -> float:
        return self._celsius

    @celsius.setter
    def celsius(self, value: float) -> None:
        if value < -273.15:
            raise ValueError(f"Below absolute zero: {value}")
        self._celsius = value

    @property
    def fahrenheit(self) -> float:
        # always computed fresh from the single source of truth (_celsius)
        return self._celsius * 9 / 5 + 32


t = Temperature(100)
print(t.celsius)     # 100
print(t.fahrenheit)  # 212.0

t.celsius = 0        # calls the setter, validates the value
print(t.fahrenheit)  # 32.0
````

### 2.5 Writing Your Own Decorator

````python
def log_call(func):
    """Decorator: print the function name and its return value on every call."""
    def wrapper(*args, **kwargs):
        print(f"-> calling {func.__name__}")
        result = func(*args, **kwargs)
        print(f"<- {func.__name__} returned {result!r}")
        return result
    return wrapper


@log_call
def add(a: int, b: int) -> int:
    return a + b


answer = add(3, 4)
print(f"answer = {answer}")
````

````python
import functools


def log_call(func):
    """Decorator that preserves the wrapped function's identity."""
    @functools.wraps(func)   # copies __name__, __doc__, __module__, etc.
    def wrapper(*args, **kwargs):
        print(f"-> {func.__name__}")
        result = func(*args, **kwargs)
        return result
    return wrapper


@log_call
def compute_churn_rate(users: list) -> float:
    """Return the percentage of churned users."""
    return sum(1 for u in users if u.get("churned")) / len(users) * 100


print(compute_churn_rate.__name__)   # compute_churn_rate — correct
print(compute_churn_rate.__doc__)    # Return the percentage... — preserved
````

### 2.7 Generators — Lazy Values with `yield`

````python
def countdown(n: int):
    """Yield integers from n down to 0, one at a time."""
    while n >= 0:
        yield n       # execution pauses here; resumes on next call
        n -= 1


gen = countdown(5)
print(next(gen))   # 5 — runs until the first yield
print(next(gen))   # 4 — resumes and runs until the next yield
print(next(gen))   # 3

# A for loop calls next() for you until StopIteration is raised
for val in countdown(3):
    print(val, end=" ")
print()   # newline
````

````python
import functools
import time


def retry_on_error(max_attempts: int = 3, delay_seconds: float = 0.5):
    """
    Decorator factory: retry a function up to max_attempts times on any exception.

    Usage:  @retry_on_error(max_attempts=3, delay_seconds=1.0)
    """
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            for attempt in range(1, max_attempts + 1):
                try:
                    return func(*args, **kwargs)
                except Exception as exc:
                    if attempt == max_attempts:
                        raise   # final attempt: let the exception propagate
                    print(
                        f"[retry] {func.__name__} attempt {attempt} failed: {exc}. "
                        f"Retrying in {delay_seconds}s..."
                    )
                    time.sleep(delay_seconds)
        return wrapper
    return decorator


# Demo: a function that fails twice, then succeeds on the third attempt
_tracker = {"count": 0}

@retry_on_error(max_attempts=3, delay_seconds=0)
def flaky_fetch(item_id: int) -> dict:
    """Simulates an API call that occasionally times out."""
    _tracker["count"] += 1
    if _tracker["count"] < 3:
        raise ConnectionError("Simulated network timeout")
    return {"id": item_id, "status": "ok"}


result = flaky_fetch(99)
print(result)
print(f"Total attempts: {_tracker['count']}")
````

````python
class BaseAPIClient:
    """Connection configuration shared by all CinemaStream API clients."""

    def __init__(self, base_url: str, token: str):
        self._base_url = base_url
        self._token = token        # private: no public .token attribute

    @property
    def headers(self) -> dict:
        """Compute auth headers on access — token never stored in a plain dict."""
        return {
            "Authorization": f"Bearer {self._token}",
            "Content-Type": "application/json",
        }

    def __repr__(self) -> str:
        # NEVER include the token in repr — it would end up in logs and crash reports
        return f"{self.__class__.__name__}(base_url='{self._base_url}', token='***')"


client = BaseAPIClient("https://api.cinemastream.sg/v1", "secret-token-abc123")
print(client)
print(client.headers)
````

### 2.1 The Context Manager Protocol: `__enter__` and `__exit__`

````python
class Resource:
    """A toy resource that must be opened and always closed."""

    def __init__(self, name: str):
        self.name = name
        self.is_open = False

    def __enter__(self):
        self.is_open = True
        print(f"opened {self.name}")
        return self                      # this is what 'as r' receives

    def __exit__(self, exc_type, exc_value, traceback) -> bool:
        self.is_open = False
        print(f"closed {self.name}")
        return False                     # False = do not suppress exceptions


with Resource("buffer") as r:
    print(f"using {r.name}, is_open={r.is_open}")

print(f"after block, is_open={r.is_open}")
````

````python
class Resource:
    def __init__(self, name: str):
        self.name = name

    def __enter__(self):
        print(f"opened {self.name}")
        return self

    def __exit__(self, exc_type, exc_value, traceback) -> bool:
        print(f"closed {self.name}")
        if exc_type is not None:
            print(f"  (cleanup ran despite {exc_type.__name__})")
        return False                     # propagate the exception to the caller


try:
    with Resource("buffer") as r:
        print("about to fail")
        raise ValueError("boom")
except ValueError as exc:
    print(f"caught: {exc}")
````

### 2.3 The Iterator Protocol: `__iter__` and `__next__`

````python
class CountUp:
    """An iterator that yields 0, 1, ..., limit-1."""

    def __init__(self, limit: int):
        self.limit = limit
        self.current = 0

    def __iter__(self):
        return self                      # an iterator returns itself

    def __next__(self) -> int:
        if self.current >= self.limit:
            raise StopIteration          # the signal that means "done"
        value = self.current
        self.current += 1
        return value


for n in CountUp(3):
    print(n)
````

### 2.4 Iterable vs Iterator — A Distinction That Bites

````python
class NumberBox:
    """An ITERABLE container — re-iterable, because __iter__ returns a fresh iterator."""

    def __init__(self, numbers: list):
        self._numbers = numbers

    def __iter__(self):
        return iter(self._numbers)       # a new list-iterator every time


box = NumberBox([10, 20, 30])
print(list(box))     # consume once
print(list(box))     # consume again — still works
print(max(box))      # any function that iterates works too
````

````python
class Ring:
    """A circular buffer: indexing wraps around using modulo."""

    def __init__(self, items):
        self._items = list(items)

    def __getitem__(self, index: int):
        return self._items[index % len(self._items)]   # wrap with modulo

    def __len__(self) -> int:
        return len(self._items)

    def __iter__(self):
        return iter(self._items)


r = Ring(["a", "b", "c"])
print(r[0])      # a
print(r[3])      # a again — wrapped
print(r[4])      # b
print(len(r))    # 3
print(list(r))   # ['a', 'b', 'c']
````

````python
import uuid
import logging
from contextvars import ContextVar

# A ContextVar is like a global variable, but scoped to the current thread/asyncio task
correlation_id: ContextVar[str] = ContextVar("correlation_id", default="no-id")

class CorrelationFilter(logging.Filter):
    """Injects the current correlation_id into every log record."""
    def filter(self, record):
        record.correlation_id = correlation_id.get()
        return True

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(correlation_id)s | %(levelname)s | %(message)s",
    datefmt="%H:%M:%S",
)

# Attach the filter to each handler
root_logger = logging.getLogger()
for handler in root_logger.handlers:
    handler.addFilter(CorrelationFilter())

logger = logging.getLogger("cinemastream.batch")


def run_batch(batch_name: str) -> None:
    """Simulate a batch that tags all its log lines with a unique run ID."""
    run_id = str(uuid.uuid4())[:8]  # short ID for readability
    token = correlation_id.set(run_id)  # set the ID for this thread/call
    try:
        logger.info(f"Starting batch: {batch_name}")
        logger.info("Loaded 500 rows from source")
        logger.warning("3 rows skipped — missing watch_minutes")
        logger.info(f"Batch complete: {batch_name}")
    finally:
        correlation_id.reset(token)  # always restore, even if an exception fires


run_batch("watch_events_2024-04-01")
run_batch("subscriptions_2024-04-01")
````

## 3. CinemaStream in Practice

````python
# cinemastream/scripts/logging_setup.py
"""
Reusable logging factory for all CinemaStream scripts.

Usage:
    from cinemastream.scripts.logging_setup import get_logger, new_correlation_id
    logger = get_logger(__name__)

    with new_correlation_id():
        logger.info("Batch started", rows=500)
"""

import json
import sys
import uuid
from contextlib import contextmanager
from contextvars import ContextVar

from loguru import logger as _loguru_logger

# A ContextVar stores the current correlation ID for this thread/asyncio task.
# Default is "no-id" so logs outside a batch context are still valid JSON.
_correlation_id: ContextVar[str] = ContextVar("correlation_id", default="no-id")


def _make_serializer():
    """Return a sink function that produces one JSON line per record."""
    def serialize_to_json(message):
        record = message.record
        payload = {
            "ts":      record["time"].isoformat(),
            "level":   record["level"].name,
            "event":   record["message"],
            "module":  record["name"],
            "cid":     _correlation_id.get(),   # correlation ID
        }
        # Merge any extra key=value pairs the caller passed (e.g., logger.info(..., rows=500))
        payload.update(record["extra"])
        return json.dumps(payload)
    return serialize_to_json


def setup_logging(level: str = "INFO") -> None:
    """
    Configure loguru to emit one JSON line per log record.
    Call this once at the top of your script's main() function.
    """
    _loguru_logger.remove()   # discard loguru's default coloured handler
    serializer = _make_serializer()
    def sink(message):
        print(serializer(message), file=sys.stderr)
    _loguru_logger.add(sink, level=level)


def get_logger(name: str):
    """
    Return a loguru logger bound to `name` (use __name__ in each module).
    Returns the module-level loguru logger with the name stored in context.
    """
    return _loguru_logger.bind(logger_name=name)


@contextmanager
def new_correlation_id(prefix: str = ""):
    """
    Context manager: set a fresh UUID correlation ID for the duration of a block.

        with new_correlation_id("ingest"):
            logger.info("Batch started")   # every line in here carries the same ID
    """
    cid = f"{prefix}-{uuid.uuid4().hex[:8]}" if prefix else uuid.uuid4().hex[:8]
    token = _correlation_id.set(cid)
    try:
        yield cid
    finally:
        _correlation_id.reset(token)
````

````python
# cinemastream/scripts/load_orders.py
"""
Load and clean subscription order data from a raw CSV export.
Writes validated rows to data/clean_orders.csv and logs rejected rows.

Usage:
    python scripts/load_orders.py
"""

import csv
import json
from pathlib import Path
from datetime import datetime


# ---- Configuration ----
DATA_DIR      = Path(__file__).parent.parent / "data"
INPUT_FILE    = DATA_DIR / "raw_orders.csv"
OUTPUT_FILE   = DATA_DIR / "clean_orders.csv"
REJECTED_FILE = DATA_DIR / "rejected_orders.json"

VALID_PLANS      = {"Free", "Basic", "Premium"}
VALID_CURRENCIES = {"SGD", "MYR", "IDR", "PHP", "THB", "VND", "INR"}


# ---- Validation ----
class OrderValidationError(ValueError):
    """Raised when an order row fails business-rule validation."""
    pass


def validate_order(row: dict) -> dict:
    """
    Validate and coerce a raw order row.
    Returns a cleaned dict. Raises OrderValidationError on failure.
    """
    try:
        order_id = int(row["order_id"])
        user_id  = int(row["user_id"])
    except (ValueError, KeyError) as e:
        raise OrderValidationError(f"Bad ID fields: {e}")

    plan = row.get("plan", "").strip()
    if plan not in VALID_PLANS:
        raise OrderValidationError(f"Unknown plan '{plan}'")

    currency = row.get("currency", "").strip().upper()
    if currency not in VALID_CURRENCIES:
        raise OrderValidationError(f"Unknown currency '{currency}'")

    try:
        amount = float(row["amount_local"])
        if amount < 0:
            raise OrderValidationError(f"Negative amount: {amount}")
    except (ValueError, KeyError) as e:
        raise OrderValidationError(f"Bad amount: {e}")

    return {
        "order_id":     order_id,
        "user_id":      user_id,
        "plan":         plan,
        "currency":     currency,
        "amount_local": amount,
    }


# ---- Main ----
def load_orders(input_path: Path, output_path: Path, rejected_path: Path) -> None:
    """Load raw orders, validate, write clean and rejected files."""
    clean_rows = []
    rejected   = []

    if not input_path.exists():
        print(f"Input file not found: {input_path}")
        return

    with open(input_path, "r", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        for row in reader:
            try:
                clean_rows.append(validate_order(row))
            except OrderValidationError as e:
                rejected.append({"row": dict(row), "reason": str(e)})

    # Write clean CSV
    output_path.parent.mkdir(parents=True, exist_ok=True)
    if clean_rows:
        fieldnames = list(clean_rows[0].keys())
        with open(output_path, "w", newline="", encoding="utf-8") as f:
            writer = csv.DictWriter(f, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(clean_rows)

    # Write rejected JSON log
    if rejected:
        with open(rejected_path, "w", encoding="utf-8") as f:
            json.dump(rejected, f, indent=2)

    print(f"Loaded {len(clean_rows)} valid orders, {len(rejected)} rejected")
    print(f"Clean:    {output_path}")
    if rejected:
        print(f"Rejected: {rejected_path}")


if __name__ == "__main__":
    load_orders(INPUT_FILE, OUTPUT_FILE, REJECTED_FILE)
````

````python
import json
from pathlib import Path

def load_and_validate_config(path: str) -> dict:
    """Load and validate a JSON pipeline config file."""
    with open(path, encoding="utf-8") as f:
        config = json.load(f)

    required = ["input_file", "output_file", "threshold"]
    for key in required:
        if key not in config:
            raise KeyError(f"Config missing required key: '{key}'. Found keys: {list(config.keys())}")

    threshold = config["threshold"]
    if not isinstance(threshold, (int, float)):
        raise ValueError(f"'threshold' must be a number, got {type(threshold).__name__}")
    if not (0.0 <= threshold <= 1.0):
        raise ValueError(f"'threshold' must be between 0 and 1, got {threshold}")

    return config

import tempfile, os, json

# Test: valid config
valid = {"input_file": "data/in.csv", "output_file": "data/out.csv", "threshold": 0.8}
with tempfile.NamedTemporaryFile(mode="w", suffix=".json", delete=False, encoding="utf-8") as f:
    json.dump(valid, f)
    tmp = f.name

result = load_and_validate_config(tmp)
print("Valid config loaded:", result)
os.unlink(tmp)

# Test: missing key
try:
    load_and_validate_config.__doc__  # just confirming function exists
    bad = {"input_file": "data/in.csv"}   # missing output_file and threshold
    with tempfile.NamedTemporaryFile(mode="w", suffix=".json", delete=False, encoding="utf-8") as f:
        json.dump(bad, f)
        tmp2 = f.name
    load_and_validate_config(tmp2)
except KeyError as e:
    print(f"Caught KeyError: {e}")
finally:
    if os.path.exists(tmp2):
        os.unlink(tmp2)

# Test: bad threshold
try:
    bad2 = {"input_file": "a", "output_file": "b", "threshold": 1.5}
    with tempfile.NamedTemporaryFile(mode="w", suffix=".json", delete=False, encoding="utf-8") as f:
        json.dump(bad2, f)
        tmp3 = f.name
    load_and_validate_config(tmp3)
except ValueError as e:
    print(f"Caught ValueError: {e}")
finally:
    if os.path.exists(tmp3):
        os.unlink(tmp3)
````

````python
import copy

def merge_env_override(base_config: dict, env_overrides: dict) -> dict:
    """
    Apply dot-notation overrides to a nested config dict.
    'validation.min_amount': 0.5 sets config['validation']['min_amount'] = 0.5
    """
    result = copy.deepcopy(base_config)
    for dotted_key, value in env_overrides.items():
        parts = dotted_key.split(".", maxsplit=1)
        if len(parts) == 1:
            result[parts[0]] = value
        else:
            section, key = parts
            if section not in result:
                result[section] = {}
            result[section][key] = value
    return result

# Test
base = {
    "pipeline": {"name": "orders-ingestion"},
    "validation": {
        "min_amount": 0.01,
        "allowed_currencies": ["SGD", "MYR"],
    },
    "logging": {"level": "INFO"},
}

overrides = {
    "validation.min_amount": 0.5,   # raise threshold for staging
    "logging.level": "DEBUG",       # more verbose in staging
}

merged = merge_env_override(base, overrides)

print("Original min_amount:", base["validation"]["min_amount"])
print("Merged min_amount:", merged["validation"]["min_amount"])
print("Original logging level:", base["logging"]["level"])
print("Merged logging level:", merged["logging"]["level"])
print("Currencies unchanged:", merged["validation"]["allowed_currencies"])
````

````python
# cinemastream/scripts/user_utils.py
"""
User-related utility functions for the CinemaStream data pipeline.
Import this module from any script that works with user data.
"""

VALID_PLANS      = frozenset({"Free", "Basic", "Premium"})
VALID_CURRENCIES = frozenset({"SGD", "MYR", "IDR", "PHP", "THB", "VND", "INR"})
VALID_COUNTRIES  = frozenset({"SG", "MY", "ID", "PH", "TH", "VN", "IN"})

PLAN_PRICES_SGD: dict[str, float] = {
    "Free":    0.00,
    "Basic":   8.90,
    "Premium": 12.90,
}


def get_plan_price(plan: str) -> float:
    """Return the monthly SGD price for a plan. Returns 0.0 for unknown plans."""
    return PLAN_PRICES_SGD.get(plan, 0.0)


def is_active(user: dict) -> bool:
    """Return True if the user has not churned."""
    return not user.get("churned", True)


def user_label(user: dict) -> str:
    """Return a one-line display string for a user."""
    status = "active" if is_active(user) else "churned"
    return f"{user['name']} [{user['plan']} | {user.get('country','??')} | {status}]"


def compute_churn_rate(users: list) -> float:
    """Return churn rate as a percentage (0.0–100.0) for a list of user dicts."""
    if not users:
        return 0.0
    churned = sum(1 for u in users if u.get("churned", False))
    return (churned / len(users)) * 100


def monthly_revenue(users: list) -> float:
    """Return total monthly SGD revenue across all active users."""
    return sum(get_plan_price(u["plan"]) for u in users if is_active(u))


if __name__ == "__main__":
    # Quick self-test — runs only when file is executed directly
    sample = [
        {"user_id": 1, "name": "Ravi Kumar",       "plan": "Premium", "country": "IN", "churned": False},
        {"user_id": 2, "name": "Siti Rahman",      "plan": "Basic",   "country": "MY", "churned": False},
        {"user_id": 3, "name": "Nguyen Van Minh",  "plan": "Free",    "country": "VN", "churned": True},
    ]
    print("=== user_utils self-test ===")
    for u in sample:
        print(" ", user_label(u))
    print(f"Churn rate: {compute_churn_rate(sample):.1f}%")
    print(f"Monthly revenue: S${monthly_revenue(sample):.2f}")
````

````python
# Addition to user_utils.py

def revenue_by_plan(users: list) -> dict:
    """Return total monthly SGD revenue grouped by plan (active users only)."""
    breakdown: dict[str, float] = {}
    for user in users:
        if not is_active(user):
            continue
        plan = user["plan"]
        breakdown[plan] = breakdown.get(plan, 0.0) + get_plan_price(plan)
    return breakdown


# In another script:
# from user_utils import revenue_by_plan
users = [
    {"user_id": 1, "name": "Ravi Kumar",       "plan": "Premium", "country": "IN", "churned": False},
    {"user_id": 2, "name": "Siti Rahman",      "plan": "Basic",   "country": "MY", "churned": False},
    {"user_id": 3, "name": "Nguyen Van Minh",  "plan": "Free",    "country": "VN", "churned": True},
]

breakdown = revenue_by_plan(users)
print("Revenue by plan:")
for plan, total in breakdown.items():
    print(f"  {plan}: S${total:.2f}")
````
